```python
import hashlib

def sign_block(prev_hash: str, data: str) -> str:
    # Объединяем предыдущий хеш и данные
    block_content = prev_hash + data

    # Переводим строку в байты
    block_bytes = block_content.encode('utf-8')

    # Вычисляем SHA256-хеш
    sha256_hash = hashlib.sha256(block_bytes)

    # Получаем хеш в виде строки
    result = sha256_hash.hexdigest()

    return result


# Пример использования
prev_hash = "abc"
data = "123"

hash_result = sign_block(prev_hash, data)

print("Предыдущий хеш:", prev_hash)
print("Данные:", data)
print("Объединенная строка:", prev_hash + data)
print("SHA256:", hash_result)
```
